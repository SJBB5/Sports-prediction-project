```python
import requests
import pandas as pd
import os
from dotenv import load_dotenv
import matplotlib.pyplot as plt
# Load the .env file
load_dotenv()

API_KEY = os.getenv("API_KEY")
```

```python
url = f"https://v1.american-football.api-sports.io/games"
headers = {
    "x-apisports-key": API_KEY
}
params = {
    "league": 1,
    "season": 2023
}

games_response = requests.get(url, headers = headers, params = params)
games_data = games_response.json()['response']
```

```python
games_df = pd.json_normalize(games_data)
display(games_df.tail(10))
```
